```python
import pandas as pd

# Step 1: Define the label mapping
label_mapping = {
    0: "other",
    1: "crater",
    2: "dark dune",
    3: "slope streak",
    4: "bright dune",
    5: "impact ejecta",
    6: "swiss cheese",
    7: "spider"
}

# Step 2: Read image names from the text file
# Update 'image_names.txt' to your actual file path
with open('hirise-map-proj-v3\labels-map-proj-v3.txt', 'r') as file:
    image_lines = file.readlines()

# Step 3: Create a list to hold the data
data = []

for line in image_lines:
    parts = line.strip().split()  # Split by whitespace
    if len(parts) > 1:  # Check if there's a label
        image_name = parts[0]  # Get the image name
        label_id = int(parts[1])  # Get the label ID
        label_name = label_mapping.get(label_id, "unknown")  # Get the corresponding label name
        data.append([image_name, label_name])  # Append the image name and label name to the data

# Step 4: Create a DataFrame and save it to a CSV file
df = pd.DataFrame(data, columns=["Image Name", "Label"])
df.to_csv('labeled_images.csv', index=False)
```
